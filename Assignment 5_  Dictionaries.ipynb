```python
highest_salary = float('-inf')
lowest_salary = float('inf')
total_salary = 0
count = 0
highest_emp = None
lowest_emp = None

for empid, details in employees.items():
    salary = details["Salary"]
    total_salary += salary
    count += 1

    if salary > highest_salary:
        highest_salary = salary
        highest_emp = (empid, details)

    if salary < lowest_salary:
        lowest_salary = salary
        lowest_emp = (empid, details)

avg_salary = total_salary / count

print("Highest salary:", highest_emp)
print("Lowest salary:", lowest_emp)
print("Average salary:", avg_salary)
```
